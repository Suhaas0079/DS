Create a dictionary, my_dictionary, which recreates the following key-value pairings:

* label = ["cat", "dog", "fish"]
* count = [2, 5, 10]

```python
my_dictionary={}
label = ["cat", "dog", "fish"]
count = [2, 5, 10]
for counter in range(len(label)):
  animal = label[counter]
  my_dictionary[animal] = count[counter]
```

### Create two randomly initialized 2D numpy arrays of integers with size 2 x 3 and 3 x 4 respectively. Hint: numpy.random.randint will provide the functionality you require here

```python
array1 = numpy.random.randint(
            100,
            dtype=int,
            size=(2,3)
            )

print(array1)
```

```python
array2= numpy.random.randint(
          100,
          dtype=int,
          size=(3, 4)
          )

print(array2)
```

### Perform a matrix multiplication of these two matrices, storing the result to a variable and printing the result.


```python
array1_Columns = array1.shape[-1]
array1_Rows = array1.shape[-2]

array2_Columns = array2.shape[-1]
array2_Rows = array2.shape[-2]

productMatrixLooped = numpy.zeros(
                        (array1_Rows, array2_Columns),
                        dtype=int
                        )

for i in range(array1_Rows):
  for j in range(array2_Columns):
    for k in range(array1_Columns):
      productMatrixLooped[i][j] += array1[i][k] * array2[k][j]

print(productMatrixLooped)
```

```python
productMatrix = numpy.dot(array1, array2)
print(productMatrix)
```

### Print only the first column of the result matrix from the last subtask.

```python
print([k[0] for k in productMatrix])
```
