### Write a Python program to read an entire text file.

```python
#Saving the file path, generally done on a .env file
path = 'E:/Repos/DS/Technicals/Python/Notebook/Support Resources/FileHandling.txt'

#Object to access the file, 'r' represents read mode
data = open(path, 'r', encoding="utf-8")

#read() returns the entire text in the file
print(type(data))

txt = data.read()
print(type(txt))

print(txt)
print("-"*100)

#Use repr to see the raw text
print(repr(txt))

#Close file and release resources
data.close()
```

#### Write a Python program to read first n lines of a file

```python
path = 'E:/Repos/DS/Technicals/Python/Notebook/Support Resources/FileHandling.txt'

data = open(path, 'r', encoding="utf-8")

text = data.readlines()

#Method1: Using readlines() and list splicing
no_of_lines = int(input("Enter no of lines to read"))
print(text[:no_of_lines])
print("-"*50)

#Variable initialisation
i=0
#Sets the cursor to the start
data.seek(0, 0)
#Method2: Using read() and looping as curson continues where left of prviously
while i<no_of_lines:
    print(data.readline())
    i+=1

#Close file and release resources
data.close()
```

#### Text preprocing using replace

```python
path2 = 'E:/Repos/DS/Technicals/Python/Notebook/Support Resources/FileHandling2.txt'

#Open file in write mode if not available create in the path
#If it is an existing file older data is cleared, if you want to add instead of restart use mode:'a' or 'a+'
data = open(path2, 'a+', encoding="utf-8")

data.seek(0)
txt = data.read()
print(txt)

#This will replace but the value returned is not stored and considered 
#Hence, the original string remains making the operation useless
txt.replace('i','')

#To resolve the above issue we replace the value in the variable
"""
    Note we have only replace the value in the variable not on the actual file. 
    That is another opertaion to be done if necssary
"""
txt = txt.replace('a','')

print(txt)

#Close file and release resources
data.close()
```
