## Data cleaning

- Here we have a total of 10841 rows with 13 columns.
- There are null values present in columns nearly 4 columns with major null values present in rating column. 
- Either handle the null values or remove the rows or remove the column based on the requirments  
- The 'rating' column has the highest null values either remove the column or hand

*Implemented soln*
- Since this is for practice, I will be removing the rows where null values are present

```python
# Store the data in new variable so we dont modify the original data variable
cleaned_data = playstore_data.dropna()

#Verify the data cleansing
print(cleaned_data.shape)
print(cleaned_data.isnull().sum())
```

## Data analysis

### Average rating for all the applications

```python
sum_of_ratings = cleaned_data['Rating'].values.sum()
application_count = cleaned_data.shape[0]

print(f"Average ratings: {round(sum_of_ratings/application_count, 3)}")
```

```python
cleaned_data['Rating'].describe(include='all')
```

### Average rating and application count for all the applications by gener

```python
#Get all possible unique values, reduces complexity for further steps
geners = cleaned_data['Genres'].unique()

#Spliting as the string has ; for seperating multiple geners
array = [gener.split(';') for gener in geners]

# Solution 1: Using itertools library
unique_geners1_secondary = set(itertools.chain.from_iterable(array))

# Solution 2: using list comprehension
unique_geners2_secondary = set([item
                                for sublist in array 
                                for item in sublist
                                ])

#Secondary filter to further simplify geners 
primary_gener_list = [primary_gener 
                for primary_gener in unique_geners1_secondary 
                if '&' not in primary_gener
                ]

secondary_gener_list = [secondary_gener 
                  for secondary_gener in unique_geners1_secondary 
                  if '&' in secondary_gener
                  ]

for secondary_gener in secondary_gener_list: 
    flag = 0
    words = secondary_gener.split('&')
    for word in words:
        if str.strip(word) not in primary_gener_list:
            flag = 1
            break

    if flag:
        primary_gener_list.append(secondary_gener)

rating_by_gener = {}
applications_by_gener = {}

for gener in sorted(primary_gener_list):
    filtered_df = cleaned_data[cleaned_data['Genres'].str.contains(gener, na=False)]
    
    applications_by_gener[gener] = len(filtered_df)
    rating_by_gener[gener] = round(
                                (sum(filtered_df['Rating']) /
                                 applications_by_gener[gener]),
                                 3)

    

custom_df = pd.DataFrame({
    'Gener': list(rating_by_gener.keys()),  # The keys (genre names) go into the 'Gener' column
    'Rating': list(rating_by_gener.values()),  # The values (ratings) go into the 'Rating' column
    'Application Count': list(applications_by_gener.values())  # The application counts go into the 'Application Count' column
})

print(custom_df)
```
